### Numpy 

```python
# np.array
import numpy as np

a = np.array([1,2,3])
print(a)
```

```python
# 2D and 3D
b = np.array([[1,2,3],[4,5,6]])
print(b)
```

```python
c = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print(c)
```

```python
# dtype
np.array([1,2,3],dtype=float)
```

```python
# np.arange
np.arange(1,20,2)
```

```python
# np.ones and np.zeros
np.ones((3,4))
```

```python
np.zeros((3,4))
```

```python
np.random.random((3,4))
```

```python
# np.identity
np.identity(3)
```

### Array Attributes

```python
a1 = np.arange(10,dtype=np.int32)
a2 = np.arange(12,dtype=float).reshape(3,4)
a3 = np.arange(8).reshape(2,2,2)
```

```python
a1 , a2 ,a3
```

```python
a1.ndim , a2.ndim , a3.ndim 
```

```python
a1.shape , a2.shape , a3.shape 
```

### Array Operations

```python
a1 = np.arange(12).reshape(3,4)
a2 = np.arange(12,24).reshape(3,4)

a1
```

```python
a1 ** 2
```

```python
a1 ** a2
```

```python
np.var(a1,axis=1)
```

```python
np.sin(a1)
```

```python
np.dot(a2,a3)
```
